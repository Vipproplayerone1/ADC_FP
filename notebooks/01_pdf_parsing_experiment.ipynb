# 01 — PDF Parsing Experiment

Compare PyMuPDF vs pypdf on the demo lecture PDFs. Verify that page-level metadata is preserved.


```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from app.services.pdf_loader import load_pdf_pages, _load_with_pymupdf, _load_with_pypdf
```

```python
PDF_DIR = Path('../data/raw/uploaded_pdfs')
pdfs = sorted(PDF_DIR.glob('*.pdf'))
for p in pdfs:
    print(p.name)
```

```python
# Compare extractors on the first PDF
target = pdfs[0]
mupdf_pages = _load_with_pymupdf(target)
pypdf_pages = _load_with_pypdf(target)

print(f'PyMuPDF pages: {len(mupdf_pages)}')
print(f'pypdf pages:   {len(pypdf_pages)}')

print('\n--- PyMuPDF page 1 (first 200 chars) ---')
print(mupdf_pages[0].text[:200])
print('\n--- pypdf page 1 (first 200 chars) ---')
print(pypdf_pages[0].text[:200])
```

```python
# Verify every page carries the required metadata
pages = load_pdf_pages(target)
for p in pages[:3]:
    print(f'{p.file_name} page {p.page_number} | {len(p.text)} chars')
    assert p.file_name == target.name
    assert p.page_number > 0
    assert p.text
```

**Conclusion**: PyMuPDF is the primary extractor; pypdf is the fallback. Both produce per-page records with the same `(file_name, page_number, text)` shape, which is what the chunker downstream relies on.
